```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
from scipy.optimize import curve_fit

# --- 日本語フォント設定 ---
font_path = "C:/Windows/Fonts/meiryo.ttc" # Windowsの場合の例

try:
    font_prop = fm.FontProperties(fname=font_path, size=12)
    plt.rcParams['font.family'] = font_prop.get_name()
    plt.rcParams['axes.unicode_minus'] = False # マイナス記号の表示対応
except FileNotFoundError:
    print(f"警告: 指定されたフォントファイルが見つかりません。{font_path}")
    print("日本語表示が正しく行われない可能性があります。")
    pass

# --- グラフ1: 実測データの準備 ---
data = [(10, 0.007895), (100, 0.009375),(1000, 0.012121),(10000, 0.056667),(20000, 0.162931),(21000, 0.258077),(22000, 0.399457),(23000, 0.583333),(24000, 0.608696),(25000, 0.613636),
        (26000, 0.514286),(27000, 0.362745),(28000, 0.253472), (29000, 0.194595), (30000, 0.147826), (31000, 0.133871), (32069, 0.112857), (33000, 0.1), (34063, 0.084375), (35000, 0.078676),
        (36000, 0.08),(37000, 0.077885), (38000, 0.074605), (39000, 0.070707), (40000, 0.065223), (50000, 0.079245), (60000, 0.032258), (70000, 0.024854), (80000, 0.024381), (90000, 0.020841),
        (100000, 0.018519),(500000, 0.00708),(1000000, 0.005941)]

exp_frequencies = np.array([item[0] for item in data]) # numpy配列に変換
exp_gains = np.array([item[1] for item in data])     # numpy配列に変換

# --- グラフ2: 理論値データの準備 (RLC直列共振回路の電流特性) ---
# 定数の定義
R_external = 51.0      # 外部抵抗 [Ω]
R_L_inductor = 29.3    # インダクタンスの抵抗 [Ω]
R_total = R_external + R_L_inductor # 総抵抗 [Ω]

L = 4700e-6 # インダクタンス [H] (4.7 mH)
C = 0.01e-6 # キャパシタンス [F] (0.01 uF)
V = 1.0     # 印加電圧 [V]

# 理論共振周波数f0を計算
f0_theory = 1 / (2 * np.pi * np.sqrt(L * C))

# 理論Q値の計算 (総抵抗を使用)
Q_theory = (2 * np.pi * f0_theory * L) / R_total

print(f"総抵抗 R_total: {R_total:.2f} Ω")
print(f"理論的な共振周波数 f0: {f0_theory:.2f} Hz")
print(f"理論的なQ値: {Q_theory:.2f}")

# 周波数の範囲を生成
min_freq_val = min(np.min(exp_frequencies), f0_theory / 100)
max_freq_val = max(np.max(exp_frequencies), f0_theory * 100)
theory_frequencies = np.logspace(np.log10(min_freq_val), np.log10(max_freq_val), 1000)

# 各周波数でのインピーダンスと電流の計算
theory_currents = []
for f in theory_frequencies:
    omega = 2 * np.pi * f
    XL = omega * L
    XC = 1 / (omega * C)
    
    Z = np.sqrt(R_total**2 + (XL - XC)**2)
    theory_currents.append(V / Z)

max_exp_gain_raw = np.max(exp_gains) # 生の実測ゲインの最大値
max_theory_current = np.max(theory_currents)
scaled_theory_gains = np.array(theory_currents) * (max_exp_gain_raw / max_theory_current)

# --- 理論値の半値全幅 (-3dB点) の計算 (前回と同様) ---
half_power_gain_theory = np.max(scaled_theory_gains) / np.sqrt(2)

peak_idx_theory = np.argmax(scaled_theory_gains)
f_low_theory = np.nan
f_high_theory = np.nan

for i in range(peak_idx_theory -1, -1, -1):
    if scaled_theory_gains[i] <= half_power_gain_theory:
        if i > 0 and scaled_theory_gains[i+1] > half_power_gain_theory:
            f_low_theory = theory_frequencies[i] + (theory_frequencies[i+1] - theory_frequencies[i]) * \
                    (half_power_gain_theory - scaled_theory_gains[i]) / (scaled_theory_gains[i+1] - scaled_theory_gains[i])
            break
        elif i == 0:
            f_low_theory = theory_frequencies[i]
            break

for i in range(peak_idx_theory + 1, len(scaled_theory_gains)):
    if scaled_theory_gains[i] <= half_power_gain_theory:
        if i < len(scaled_theory_gains) -1 and scaled_theory_gains[i-1] > half_power_gain_theory:
            f_high_theory = theory_frequencies[i-1] + (theory_frequencies[i] - theory_frequencies[i-1]) * \
                     (half_power_gain_theory - scaled_theory_gains[i-1]) / (scaled_theory_gains[i] - scaled_theory_gains[i-1])
            break
        elif i == len(scaled_theory_gains) -1:
            f_high_theory = theory_frequencies[i]
            break

delta_f_theory = f_high_theory - f_low_theory if not np.isnan(f_low_theory) and not np.isnan(f_high_theory) else np.nan
print(f"理論的な帯域幅 (Δf_theory): {delta_f_theory:.2f} Hz")


# --- 実測データのフィット (ローレンツ関数) ---

def lorentzian(x, A, x0, gamma, B):
    """
    ローレンツ関数: A * (gamma/2)^2 / ((x - x0)^2 + (gamma/2)^2) + B
    gamma は半値全幅
    """
    return A * ((gamma / 2)**2) / ((x - x0)**2 + (gamma / 2)**2) + B

# 初期パラメータの自動推定の改善
exp_peak_idx_for_fit = np.argmax(exp_gains)
initial_x0_fit = exp_frequencies[exp_peak_idx_for_fit]
initial_max_gain = exp_gains[exp_peak_idx_for_fit]
initial_min_gain = np.min(exp_gains)
initial_A_fit = initial_max_gain - initial_min_gain # ピーク高さの推定
initial_B_fit = initial_min_gain # ベースラインの推定

# 半値全幅の初期推定を、データからもう少し賢く行う
half_gain_for_gamma = initial_max_gain - initial_A_fit / 2 # Aがピーク高さなので、A/2の高さの点を探す
low_freq_gamma_idx = np.where(exp_frequencies < initial_x0_fit)[0]
high_freq_gamma_idx = np.where(exp_frequencies > initial_x0_fit)[0]

f_low_gamma_est = np.nan
if len(low_freq_gamma_idx) > 0:
    for i in low_freq_gamma_idx[::-1]: # ピークから左に
        if exp_gains[i] <= half_gain_for_gamma:
            # 簡易線形補間
            if i + 1 < len(exp_frequencies) and exp_gains[i+1] > half_gain_for_gamma:
                 f_low_gamma_est = exp_frequencies[i] + (exp_frequencies[i+1] - exp_frequencies[i]) * \
                                   (half_gain_for_gamma - exp_gains[i]) / (exp_gains[i+1] - exp_gains[i])
            else:
                 f_low_gamma_est = exp_frequencies[i]
            break
if np.isnan(f_low_gamma_est):
    f_low_gamma_est = exp_frequencies[0]

f_high_gamma_est = np.nan
if len(high_freq_gamma_idx) > 0:
    for i in high_freq_gamma_idx: # ピークから右に
        if exp_gains[i] <= half_gain_for_gamma:
            # 簡易線形補間
            if i > 0 and exp_gains[i-1] > half_gain_for_gamma:
                f_high_gamma_est = exp_frequencies[i-1] + (exp_frequencies[i] - exp_frequencies[i-1]) * \
                                   (half_gain_for_gamma - exp_gains[i-1]) / (exp_gains[i] - exp_gains[i-1])
            else:
                f_high_gamma_est = exp_frequencies[i]
            break
if np.isnan(f_high_gamma_est):
    f_high_gamma_est = exp_frequencies[-1]

initial_gamma_fit = abs(f_high_gamma_est - f_low_gamma_est) if not np.isnan(f_low_gamma_est) and not np.isnan(f_high_gamma_est) else 5000 # デフォルト値

p0 = [initial_A_fit, initial_x0_fit, initial_gamma_fit, initial_B_fit]

# フィット範囲の限定 (共振周波数の0.5倍から2倍の範囲に限定)
# データ点が多い場合は、ピークの左右の半値点±αの範囲にする
fit_min_freq = initial_x0_fit * 0.5
fit_max_freq = initial_x0_fit * 2.0

fit_indices = np.where((exp_frequencies >= fit_min_freq) & (exp_frequencies <= fit_max_freq))
exp_frequencies_fit_range = exp_frequencies[fit_indices]
exp_gains_fit_range = exp_gains[fit_indices]

# フィットの実行
fit_success = False
if len(exp_frequencies_fit_range) > 3: # 最低4つのパラメータなので4点以上必要
    try:
        # boundsでパラメータの物理的な範囲を指定することで、より安定したフィットが可能
        # A: 0より大きい, x0: ピーク周波数付近, gamma: 0より大きい, B: 0以上
        bounds = ([0, initial_x0_fit * 0.5, 0, 0], [np.inf, initial_x0_fit * 2, np.inf, np.max(exp_gains)])
        popt, pcov = curve_fit(lorentzian, exp_frequencies_fit_range, exp_gains_fit_range, p0=p0, bounds=bounds, maxfev=10000)
        # 最適化されたパラメータ
        fit_A, fit_x0, fit_gamma, fit_B = popt

        # フィット曲線上のデータを生成
        # フィット範囲だけでなく、広い範囲で滑らかにプロット
        plot_fit_frequencies = np.logspace(np.log10(min_freq_val), np.log10(max_freq_val), 500)
        fit_gains_for_plot = lorentzian(plot_fit_frequencies, fit_A, fit_x0, fit_gamma, fit_B)

        # フィット結果からの実測Q値と帯域幅の計算
        f0_exp_fit = fit_x0
        delta_f_exp_fit = fit_gamma # ローレンツ関数のgammaが半値全幅
        Q_exp_fit = f0_exp_fit / delta_f_exp_fit if delta_f_exp_fit > 0 else np.nan

        print(f"\n--- 実測データ フィット結果 ---")
        print(f"初期パラメータ (推定): A={initial_A_fit:.4f}, x0={initial_x0_fit:.2f}, gamma={initial_gamma_fit:.2f}, B={initial_B_fit:.4f}")
        print(f"フィットされたパラメータ: A={fit_A:.4f}, x0={fit_x0:.2f}, gamma={fit_gamma:.2f}, B={fit_B:.4f}")
        print(f"フィットされた共振周波数 (f0_exp_fit): {f0_exp_fit:.2f} Hz")
        print(f"フィットされた帯域幅 (Δf_exp_fit): {delta_f_exp_fit:.2f} Hz")
        print(f"フィットされたQ値 (Q_exp_fit): {Q_exp_fit:.2f}")

        fit_success = True
    except (RuntimeError, ValueError) as e:
        print(f"\n--- 実測データ フィット失敗 ---")
        print(f"エラー: {e}")
        print("初期パラメータの調整やフィット範囲の再検討が必要かもしれません。")

# --- グラフの描画 ---
plt.figure(figsize=(12, 7))

# グラフ1: 実測データの散布図
plt.scatter(exp_frequencies, exp_gains, color='blue', marker='o', label='測定値', zorder=5) # zorderで手前に表示

# フィット曲線があればプロット
if fit_success:
    plt.plot(plot_fit_frequencies, fit_gains_for_plot, color='purple', linestyle='--', linewidth=2, label='測定値フィット曲線', zorder=4)
    plt.axvline(x=f0_exp_fit, color='lightblue', linestyle='--', label=f'実測フィット $f_0 = {f0_exp_fit/1000:.2f}$ kHz', zorder=3) # フィットからのf0

# グラフ2: 理論値の線グラフ (スケール調整後)
plt.plot(theory_frequencies, scaled_theory_gains, color='red', linestyle='-', linewidth=2, label='理論値', zorder=2)

# --- テキスト表示位置の調整 ---
x_right_pos = max_freq_val * 0.9
x_left_pos = min_freq_val * 1.5

# 理論共振周波数の値を右に寄せる
plt.text(x_right_pos, np.max(scaled_theory_gains) * 0.8, f'理論 $f_0$={f0_theory/1000:.2f}kHz',
         color='gray', ha='right', va='center', fontproperties=font_prop)

# 理論Q値のテキストを右に寄せる
plt.text(x_right_pos, np.max(scaled_theory_gains) * 0.95, f'理論Q値: {Q_theory:.2f}',
         color='darkviolet', ha='right', va='top', fontproperties=font_prop,
         bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="darkviolet", lw=0.5, alpha=0.8))

# 実測フィットQ値のテキストを左に寄せる
if fit_success and not np.isnan(Q_exp_fit):
    plt.text(x_left_pos, np.max(exp_gains) * 0.85, f'実測フィットQ値: {Q_exp_fit:.2f}',
             color='darkblue', ha='left', va='top', fontproperties=font_prop,
             bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="darkblue", lw=0.5, alpha=0.8))
    # 実測フィット共振周波数も表示
    plt.text(x_left_pos, np.max(exp_gains) * 0.7, f'実測フィット $f_0$={f0_exp_fit/1000:.2f}kHz',
             color='blue', ha='left', va='center', fontproperties=font_prop)


# 帯域幅の表示
# 理論値の帯域幅
if not np.isnan(delta_f_theory):
    plt.hlines(y=half_power_gain_theory, xmin=f_low_theory, xmax=f_high_theory, color='green', linestyle=':', label='理論-3dB帯域幅', zorder=1)
    plt.text((f_low_theory + f_high_theory) / 2, half_power_gain_theory * 1.1, f'理論$\\Delta f$={delta_f_theory/1000:.2f}kHz',
             color='green', ha='center', va='bottom', fontproperties=font_prop)

# 実測フィットからの帯域幅 (半値点とライン)
if fit_success and not np.isnan(delta_f_exp_fit):
    # ローレンツ関数のgammaが半値全幅なので、これを使って半値点の高さを計算
    # 半値点は A/2 + B の高さ
    half_power_gain_fit_val = fit_A / 2 + fit_B
    f_low_fit = fit_x0 - fit_gamma / 2
    f_high_fit = fit_x0 + fit_gamma / 2

    plt.hlines(y=half_power_gain_fit_val, xmin=f_low_fit, xmax=f_high_fit, color='darkorange', linestyle='-.', label='実測フィット-3dB帯域幅', zorder=1)
    plt.text((f_low_fit + f_high_fit) / 2, half_power_gain_fit_val * 0.9, f'実測フィット$\\Delta f$={delta_f_exp_fit/1000:.2f}kHz',
             color='darkorange', ha='center', va='top', fontproperties=font_prop)


plt.xscale('log') # 周波数は対数スケールで表示
plt.xlabel("周波数 [Hz]", fontproperties=font_prop)
plt.ylabel("ゲイン (相対値)", fontproperties=font_prop) # ゲインとして統一
plt.title("実測ゲインとRLC直列共振回路の理論ゲインの比較 (フィット曲線とQ値表示)", fontproperties=font_prop)
plt.grid(True, which="both", ls="--", c='0.7') # 主グリッドと副グリッドの両方を表示

# y軸の範囲を調整 (両方のデータが収まるように)
max_y_value = max(np.max(exp_gains), np.max(scaled_theory_gains))
plt.ylim(0, max_y_value * 1.1)

plt.legend(prop=font_prop, loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
from scipy.optimize import curve_fit

# --- 日本語フォント設定 ---
font_path = "C:/Windows/Fonts/meiryo.ttc" # Windowsの場合の例

try:
    font_prop = fm.FontProperties(fname=font_path, size=20) # フォントサイズを20に統一
    plt.rcParams['font.family'] = font_prop.get_name()
    plt.rcParams['axes.unicode_minus'] = False # マイナス記号の表示対応
except FileNotFoundError:
    print(f"警告: 指定されたフォントファイルが見つかりません。{font_path}")
    print("日本語表示が正しく行われない可能性があります。")
    plt.rcParams['font.family'] = 'sans-serif'
    plt.rcParams['font.size'] = 20 # デフォルトフォントサイズも20に設定
    plt.rcParams['axes.unicode_minus'] = False
    pass

# --- グラフ1: 実測データの準備 ---
data = [(10, 0.007895), (100, 0.009375),(1000, 0.012121),(10000, 0.056667),(20000, 0.162931),(21000, 0.258077),(22000, 0.399457),(23000, 0.583333),(24000, 0.608696),(25000, 0.613636),
        (26000, 0.514286),(27000, 0.362745),(28000, 0.253472), (29000, 0.194595), (30000, 0.147826), (31000, 0.133871), (32069, 0.112857), (33000, 0.1), (34063, 0.084375), (35000, 0.078676),
        (36000, 0.08),(37000, 0.077885), (38000, 0.074605), (39000, 0.070707), (40000, 0.065223), (50000, 0.079245), (60000, 0.032258), (70000, 0.024854), (80000, 0.024381), (90000, 0.020841),
        (100000, 0.018519),(500000, 0.00708),(1000000, 0.005941)]

exp_frequencies = np.array([item[0] for item in data]) # numpy配列に変換
exp_gains = np.array([item[1] for item in data])     # numpy配列に変換

# --- グラフ2: 理論値データの準備 (RLC直列共振回路の電流特性) ---
# 定数の定義
R_external = 51.0      # 外部抵抗 [Ω]
R_L_inductor = 29.3    # インダクタンスの抵抗 [Ω]
R_total = R_external + R_L_inductor # 総抵抗 [Ω]

L = 4700e-6 # インダクタンス [H] (4.7 mH)
C = 0.01e-6 # キャパシタンス [F] (0.01 uF)
V = 1.0     # 印加電圧 [V]

# 理論共振周波数f0を計算
f0_theory = 1 / (2 * np.pi * np.sqrt(L * C))

# 理論Q値の計算 (総抵抗を使用)
Q_theory = (2 * np.pi * f0_theory * L) / R_total

print(f"総抵抗 R_total: {R_total:.2f} Ω")
print(f"理論的な共振周波数 f0: {f0_theory:.2f} Hz")
print(f"理論的なQ値: {Q_theory:.2f}")

# 周波数の範囲を生成
min_freq_val = min(np.min(exp_frequencies), f0_theory / 100)
max_freq_val = max(np.max(exp_frequencies), f0_theory * 100)
theory_frequencies = np.logspace(np.log10(min_freq_val), np.log10(max_freq_val), 1000)

# 各周波数でのインピーダンスと電流の計算
theory_currents = []
for f in theory_frequencies:
    omega = 2 * np.pi * f
    XL = omega * L
    XC = 1 / (omega * C)
    
    Z = np.sqrt(R_total**2 + (XL - XC)**2)
    theory_currents.append(V / Z)

max_exp_gain_raw = np.max(exp_gains) # 生の実測ゲインの最大値
max_theory_current = np.max(theory_currents)
scaled_theory_gains = np.array(theory_currents) * (max_exp_gain_raw / max_theory_current)

# --- 理論値の半値全幅 (-3dB点) の計算 (前回と同様) ---
half_power_gain_theory = np.max(scaled_theory_gains) / np.sqrt(2)

peak_idx_theory = np.argmax(scaled_theory_gains)
f_low_theory = np.nan
f_high_theory = np.nan

for i in range(peak_idx_theory -1, -1, -1):
    if scaled_theory_gains[i] <= half_power_gain_theory:
        if i > 0 and scaled_theory_gains[i+1] > half_power_gain_theory:
            f_low_theory = theory_frequencies[i] + (theory_frequencies[i+1] - theory_frequencies[i]) * \
                           (half_power_gain_theory - scaled_theory_gains[i]) / (scaled_theory_gains[i+1] - scaled_theory_gains[i])
            break
        elif i == 0:
            f_low_theory = theory_frequencies[i]
            break

for i in range(peak_idx_theory + 1, len(scaled_theory_gains)):
    if scaled_theory_gains[i] <= half_power_gain_theory:
        if i < len(scaled_theory_gains) -1 and scaled_theory_gains[i-1] > half_power_gain_theory:
            f_high_theory = theory_frequencies[i-1] + (theory_frequencies[i] - theory_frequencies[i-1]) * \
                           (half_power_gain_theory - scaled_theory_gains[i-1]) / (scaled_theory_gains[i] - scaled_theory_gains[i-1])
            break
        elif i == len(scaled_theory_gains) -1:
            f_high_theory = theory_frequencies[i]
            break

delta_f_theory = f_high_theory - f_low_theory if not np.isnan(f_low_theory) and not np.isnan(f_high_theory) else np.nan
print(f"理論的な帯域幅 (Δf_theory): {delta_f_theory:.2f} Hz")


# --- 実測データのフィット (ローレンツ関数) ---

def lorentzian(x, A, x0, gamma, B):
    """
    ローレンツ関数: A * (gamma/2)^2 / ((x - x0)^2 + (gamma/2)^2) + B
    gamma は半値全幅
    """
    return A * ((gamma / 2)**2) / ((x - x0)**2 + (gamma / 2)**2) + B

# 初期パラメータの自動推定の改善
exp_peak_idx_for_fit = np.argmax(exp_gains)
initial_x0_fit = exp_frequencies[exp_peak_idx_for_fit]
initial_max_gain = exp_gains[exp_peak_idx_for_fit]
initial_min_gain = np.min(exp_gains)
initial_A_fit = initial_max_gain - initial_min_gain # ピーク高さの推定
initial_B_fit = initial_min_gain # ベースラインの推定

# 半値全幅の初期推定を、データからもう少し賢く行う
half_gain_for_gamma = initial_max_gain - initial_A_fit / 2 # Aがピーク高さなので、A/2の高さの点を探す
low_freq_gamma_idx = np.where(exp_frequencies < initial_x0_fit)[0]
high_freq_gamma_idx = np.where(exp_frequencies > initial_x0_fit)[0]

f_low_gamma_est = np.nan
if len(low_freq_gamma_idx) > 0:
    for i in low_freq_gamma_idx[::-1]: # ピークから左に
        if exp_gains[i] <= half_gain_for_gamma:
            # 簡易線形補間
            if i + 1 < len(exp_frequencies) and exp_gains[i+1] > half_gain_for_gamma:
                f_low_gamma_est = exp_frequencies[i] + (exp_frequencies[i+1] - exp_frequencies[i]) * \
                                  (half_gain_for_gamma - exp_gains[i]) / (exp_gains[i+1] - exp_gains[i])
            else:
                f_low_gamma_est = exp_frequencies[i]
            break
if np.isnan(f_low_gamma_est):
    f_low_gamma_est = exp_frequencies[0]

f_high_gamma_est = np.nan
if len(high_freq_gamma_idx) > 0:
    for i in high_freq_gamma_idx: # ピークから右に
        if exp_gains[i] <= half_gain_for_gamma:
            # 簡易線形補間
            if i > 0 and exp_gains[i-1] > half_gain_for_gamma:
                f_high_gamma_est = exp_frequencies[i-1] + (exp_frequencies[i] - exp_frequencies[i-1]) * \
                                  (half_gain_for_gamma - exp_gains[i-1]) / (exp_gains[i] - exp_gains[i-1])
            else:
                f_high_gamma_est = exp_frequencies[i]
            break
if np.isnan(f_high_gamma_est):
    f_high_gamma_est = exp_frequencies[-1]

initial_gamma_fit = abs(f_high_gamma_est - f_low_gamma_est) if not np.isnan(f_low_gamma_est) and not np.isnan(f_high_gamma_est) else 5000 # デフォルト値

p0 = [initial_A_fit, initial_x0_fit, initial_gamma_fit, initial_B_fit]

# フィット範囲の限定 (共振周波数の0.5倍から2倍の範囲に限定)
# データ点が多い場合は、ピークの左右の半値点±αの範囲にする
fit_min_freq = initial_x0_fit * 0.5
fit_max_freq = initial_x0_fit * 2.0

fit_indices = np.where((exp_frequencies >= fit_min_freq) & (exp_frequencies <= fit_max_freq))
exp_frequencies_fit_range = exp_frequencies[fit_indices]
exp_gains_fit_range = exp_gains[fit_indices]

# フィットの実行
fit_success = False
if len(exp_frequencies_fit_range) > 3: # 最低4つのパラメータなので4点以上必要
    try:
        # Bの最大値を、観測された最小ゲインの5倍に制限することで、ベースラインのフィットを改善
        bounds = ([0, initial_x0_fit * 0.5, 0, 0], [np.inf, initial_x0_fit * 2, np.inf, initial_min_gain * 5])
        popt, pcov = curve_fit(lorentzian, exp_frequencies_fit_range, exp_gains_fit_range, p0=p0, bounds=bounds, maxfev=10000)
        # 最適化されたパラメータ
        fit_A, fit_x0, fit_gamma, fit_B = popt

        # フィット曲線上のデータを生成
        plot_fit_frequencies = np.logspace(np.log10(min_freq_val), np.log10(max_freq_val), 500)
        fit_gains_for_plot = lorentzian(plot_fit_frequencies, fit_A, fit_x0, fit_gamma, fit_B)

        # フィット結果からの実測Q値と帯域幅の計算
        f0_exp_fit = fit_x0
        delta_f_exp_fit = fit_gamma # ローレンツ関数のgammaが半値全幅
        Q_exp_fit = f0_exp_fit / delta_f_exp_fit if delta_f_exp_fit > 0 else np.nan

        print(f"\n--- 実測データ フィット結果 ---")
        print(f"初期パラメータ (推定): A={initial_A_fit:.4f}, x0={initial_x0_fit:.2f}, gamma={initial_gamma_fit:.2f}, B={initial_B_fit:.4f}")
        print(f"フィットされたパラメータ: A={fit_A:.4f}, x0={fit_x0:.2f}, gamma={fit_gamma:.2f}, B={fit_B:.4f}")
        print(f"フィットされた共振周波数 (f0_exp_fit): {f0_exp_fit:.2f} Hz")
        print(f"フィットされた帯域幅 (Δf_exp_fit): {delta_f_exp_fit:.2f} Hz")
        print(f"フィットされたQ値 (Q_exp_fit): {Q_exp_fit:.2f}")

        fit_success = True
    except (RuntimeError, ValueError) as e:
        print(f"\n--- 実測データ フィット失敗 ---")
        print(f"エラー: {e}")
        print("初期パラメータの調整やフィット範囲の再検討が必要かもしれません。")

# --- グラフの描画 ---
plt.figure(figsize=(12, 7))

# グラフ1: 実測データの散布図
plt.scatter(exp_frequencies, exp_gains, color='blue', marker='o', label='測定値', zorder=5) # zorderで手前に表示

# フィット曲線があればプロット
if fit_success:
    plt.plot(plot_fit_frequencies, fit_gains_for_plot, color='purple', linestyle='--', linewidth=2, label='測定値フィット曲線', zorder=4)
    plt.axvline(x=f0_exp_fit, color='lightblue', linestyle='--', label=f'実測フィット $f_0 = {f0_exp_fit/1000:.2f}$ kHz', zorder=3) # フィットからのf0

# グラフ2: 理論値の線グラフ (スケール調整後)
plt.plot(theory_frequencies, scaled_theory_gains, color='red', linestyle='-', linewidth=2, label='理論値', zorder=2)

# --- テキスト表示位置の調整 ---
x_right_pos = max_freq_val * 0.9
x_left_pos = min_freq_val * 1.5


 # 実測フィット共振周波数も表示
plt.text(x_left_pos, np.max(exp_gains) * 0.7, f'実測フィット $f_0$={f0_exp_fit/1000:.2f}kHz',
             color='blue', ha='left', va='center', fontproperties=font_prop, fontsize=20)


# 帯域幅の表示
# 理論値の帯域幅
if not np.isnan(delta_f_theory):
    plt.hlines(y=half_power_gain_theory, xmin=f_low_theory, xmax=f_high_theory, color='green', linestyle=':', label='理論-3dB帯域幅', zorder=1)
    plt.text((f_low_theory + f_high_theory) / 2, half_power_gain_theory * 1.1, f'理論$\\Delta f$={delta_f_theory/1000:.2f}kHz',
             color='green', ha='center', va='bottom', fontproperties=font_prop, fontsize=20)

# 実測フィットからの帯域幅 (半値点とライン)
if fit_success and not np.isnan(delta_f_exp_fit):
    # ローレンツ関数のgammaが半値全幅なので、これを使って半値点の高さを計算
    # 半値点は A/2 + B の高さ
    half_power_gain_fit_val = fit_A / 2 + fit_B
    f_low_fit = fit_x0 - fit_gamma / 2
    f_high_fit = fit_x0 + fit_gamma / 2

    plt.hlines(y=half_power_gain_fit_val, xmin=f_low_fit, xmax=f_high_fit, color='darkorange', linestyle='-.', label='実測フィット-3dB帯域幅', zorder=1)
    plt.text((f_low_fit + f_high_fit) / 2, half_power_gain_fit_val * 0.9, f'実測フィット$\\Delta f$={delta_f_exp_fit/1000:.2f}kHz',
             color='darkorange', ha='center', va='top', fontproperties=font_prop, fontsize=20)


plt.xscale('log') # 周波数は対数スケールで表示
plt.xlabel("周波数 [Hz]", fontproperties=font_prop, fontsize=20)
plt.ylabel("ゲイン ", fontproperties=font_prop, fontsize=20) # ゲインとして統一
plt.title("測定値ゲインとRLC直列共振回路の理論ゲインの比較", fontproperties=font_prop, fontsize=20)
plt.grid(True, which="both", ls="--", c='0.7') # 主グリッドと副グリッドの両方を表示

# y軸の範囲を調整 (両方のデータが収まるように)
max_y_value = max(np.max(exp_gains), np.max(scaled_theory_gains))
plt.ylim(0, max_y_value * 1.1)

plt.legend(prop=font_prop, loc='upper left', fontsize=20) # 凡例のフォントサイズも20に設定
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
from scipy.optimize import curve_fit

# --- 日本語フォント設定 ---
font_path = "C:/Windows/Fonts/meiryo.ttc" # Windowsの場合の例

try:
    font_prop = fm.FontProperties(fname=font_path, size=20) # フォントサイズを20に統一
    plt.rcParams['font.family'] = font_prop.get_name()
    plt.rcParams['axes.unicode_minus'] = False # マイナス記号の表示対応
except FileNotFoundError:
    print(f"警告: 指定されたフォントファイルが見つかりません。{font_path}")
    print("日本語表示が正しく行われない可能性があります。")
    plt.rcParams['font.family'] = 'sans-serif'
    plt.rcParams['font.size'] = 20 # デフォルトフォントサイズも20に設定
    plt.rcParams['axes.unicode_minus'] = False
    pass

# --- グラフ1: 実測データの準備 ---
data = [(10, 0.007895), (100, 0.009375),(1000, 0.012121),(10000, 0.056667),(20000, 0.162931),(21000, 0.258077),(22000, 0.399457),(23000, 0.583333),(24000, 0.608696),(25000, 0.613636),
        (26000, 0.514286),(27000, 0.362745),(28000, 0.253472), (29000, 0.194595), (30000, 0.147826), (31000, 0.133871), (32069, 0.112857), (33000, 0.1), (34063, 0.084375), (35000, 0.078676),
        (36000, 0.08),(37000, 0.077885), (38000, 0.074605), (39000, 0.070707), (40000, 0.065223), (50000, 0.079245), (60000, 0.032258), (70000, 0.024854), (80000, 0.024381), (90000, 0.020841),
        (100000, 0.018519),(500000, 0.00708),(1000000, 0.005941)]

exp_frequencies = np.array([item[0] for item in data]) # numpy配列に変換
exp_gains = np.array([item[1] for item in data])     # numpy配列に変換

# --- グラフ2: 理論値データの準備 (RLC直列共振回路の電流特性) ---
# 定数の定義
R_external = 51.0      # 外部抵抗 [Ω]
R_L_inductor = 29.3    # インダクタンスの抵抗 [Ω]
R_total = R_external + R_L_inductor # 総抵抗 [Ω]

L = 4700e-6 # インダクタンス [H] (4.7 mH)
C = 0.01e-6 # キャパシタンス [F] (0.01 uF)
V = 1.0     # 印加電圧 [V]

# 理論共振周波数f0を計算
f0_theory = 1 / (2 * np.pi * np.sqrt(L * C))

# 理論Q値の計算 (総抵抗を使用)
Q_theory = (2 * np.pi * f0_theory * L) / R_total

print(f"総抵抗 R_total: {R_total:.2f} Ω")
print(f"理論的な共振周波数 f0: {f0_theory:.2f} Hz")
print(f"理論的なQ値: {Q_theory:.2f}")

# 周波数の範囲を生成
min_freq_val = min(np.min(exp_frequencies), f0_theory / 100)
max_freq_val = max(np.max(exp_frequencies), f0_theory * 100)
theory_frequencies = np.logspace(np.log10(min_freq_val), np.log10(max_freq_val), 1000)

# 各周波数でのインピーダンスと電流の計算
theory_currents = []
for f in theory_frequencies:
    omega = 2 * np.pi * f
    XL = omega * L
    XC = 1 / (omega * C)
    
    Z = np.sqrt(R_total**2 + (XL - XC)**2)
    theory_currents.append(V / Z)

max_exp_gain_raw = np.max(exp_gains) # 生の実測ゲインの最大値
max_theory_current = np.max(theory_currents)
scaled_theory_gains = np.array(theory_currents) * (max_exp_gain_raw / max_theory_current)

# --- 理論値の半値全幅 (-3dB点) の計算 (前回と同様) ---
half_power_gain_theory = np.max(scaled_theory_gains) / np.sqrt(2)

peak_idx_theory = np.argmax(scaled_theory_gains)
f_low_theory = np.nan
f_high_theory = np.nan

for i in range(peak_idx_theory -1, -1, -1):
    if scaled_theory_gains[i] <= half_power_gain_theory:
        if i > 0 and scaled_theory_gains[i+1] > half_power_gain_theory:
            f_low_theory = theory_frequencies[i] + (theory_frequencies[i+1] - theory_frequencies[i]) * \
                           (half_power_gain_theory - scaled_theory_gains[i]) / (scaled_theory_gains[i+1] - scaled_theory_gains[i])
            break
        elif i == 0:
            f_low_theory = theory_frequencies[i]
            break

for i in range(peak_idx_theory + 1, len(scaled_theory_gains)):
    if scaled_theory_gains[i] <= half_power_gain_theory:
        if i < len(scaled_theory_gains) -1 and scaled_theory_gains[i-1] > half_power_gain_theory:
            f_high_theory = theory_frequencies[i-1] + (theory_frequencies[i] - theory_frequencies[i-1]) * \
                           (half_power_gain_theory - scaled_theory_gains[i-1]) / (scaled_theory_gains[i] - scaled_theory_gains[i-1])
            break
        elif i == len(scaled_theory_gains) -1:
            f_high_theory = theory_frequencies[i]
            break

delta_f_theory = f_high_theory - f_low_theory if not np.isnan(f_low_theory) and not np.isnan(f_high_theory) else np.nan
print(f"理論的な帯域幅 (Δf_theory): {delta_f_theory:.2f} Hz")


# --- 実測データのフィット (ローレンツ関数) ---

def lorentzian(x, A, x0, gamma, B):
    """
    ローレンツ関数: A * (gamma/2)^2 / ((x - x0)^2 + (gamma/2)^2) + B
    gamma は半値全幅
    """
    return A * ((gamma / 2)**2) / ((x - x0)**2 + (gamma / 2)**2) + B

# 初期パラメータの自動推定の改善
exp_peak_idx_for_fit = np.argmax(exp_gains)
initial_x0_fit = exp_frequencies[exp_peak_idx_for_fit]
initial_max_gain = exp_gains[exp_peak_idx_for_fit]
initial_min_gain = np.min(exp_gains)
initial_A_fit = initial_max_gain - initial_min_gain # ピーク高さの推定
initial_B_fit = initial_min_gain # ベースラインの推定

# 半値全幅の初期推定を、データからもう少し賢く行う
half_gain_for_gamma = initial_max_gain - initial_A_fit / 2 # Aがピーク高さなので、A/2の高さの点を探す
low_freq_gamma_idx = np.where(exp_frequencies < initial_x0_fit)[0]
high_freq_gamma_idx = np.where(exp_frequencies > initial_x0_fit)[0]

f_low_gamma_est = np.nan
if len(low_freq_gamma_idx) > 0:
    for i in low_freq_gamma_idx[::-1]: # ピークから左に
        if exp_gains[i] <= half_gain_for_gamma:
            # 簡易線形補間
            if i + 1 < len(exp_frequencies) and exp_gains[i+1] > half_gain_for_gamma:
                f_low_gamma_est = exp_frequencies[i] + (exp_frequencies[i+1] - exp_frequencies[i]) * \
                                  (half_gain_for_gamma - exp_gains[i]) / (exp_gains[i+1] - exp_gains[i])
            else:
                f_low_gamma_est = exp_frequencies[i]
            break
if np.isnan(f_low_gamma_est):
    f_low_gamma_est = exp_frequencies[0]

f_high_gamma_est = np.nan
if len(high_freq_gamma_idx) > 0:
    for i in high_freq_gamma_idx: # ピークから右に
        if exp_gains[i] <= half_gain_for_gamma:
            # 簡易線形補間
            if i > 0 and exp_gains[i-1] > half_gain_for_gamma:
                f_high_gamma_est = exp_frequencies[i-1] + (exp_frequencies[i] - exp_frequencies[i-1]) * \
                                  (half_gain_for_gamma - exp_gains[i-1]) / (exp_gains[i] - exp_gains[i-1])
            else:
                f_high_gamma_est = exp_frequencies[i]
            break
if np.isnan(f_high_gamma_est):
    f_high_gamma_est = exp_frequencies[-1]

initial_gamma_fit = abs(f_high_gamma_est - f_low_gamma_est) if not np.isnan(f_low_gamma_est) and not np.isnan(f_high_gamma_est) else 5000 # デフォルト値

p0 = [initial_A_fit, initial_x0_fit, initial_gamma_fit, initial_B_fit]

# フィット範囲の限定 (共振周波数の0.5倍から2倍の範囲に限定)
# データ点が多い場合は、ピークの左右の半値点±αの範囲にする
fit_min_freq = initial_x0_fit * 0.5
fit_max_freq = initial_x0_fit * 2.0

fit_indices = np.where((exp_frequencies >= fit_min_freq) & (exp_frequencies <= fit_max_freq))
exp_frequencies_fit_range = exp_frequencies[fit_indices]
exp_gains_fit_range = exp_gains[fit_indices]

# フィットの実行
fit_success = False
if len(exp_frequencies_fit_range) > 3: # 最低4つのパラメータなので4点以上必要
    try:
        # Bの最大値を、観測された最小ゲインの5倍に制限することで、ベースラインのフィットを改善
        bounds = ([0, initial_x0_fit * 0.5, 0, 0], [np.inf, initial_x0_fit * 2, np.inf, initial_min_gain * 5])
        popt, pcov = curve_fit(lorentzian, exp_frequencies_fit_range, exp_gains_fit_range, p0=p0, bounds=bounds, maxfev=10000)
        # 最適化されたパラメータ
        fit_A, fit_x0, fit_gamma, fit_B = popt

        # フィット曲線上のデータを生成
        plot_fit_frequencies = np.logspace(np.log10(min_freq_val), np.log10(max_freq_val), 500)
        fit_gains_for_plot = lorentzian(plot_fit_frequencies, fit_A, fit_x0, fit_gamma, fit_B)

        # フィット結果からの実測Q値と帯域幅の計算
        f0_exp_fit = fit_x0
        delta_f_exp_fit = fit_gamma # ローレンツ関数のgammaが半値全幅
        Q_exp_fit = f0_exp_fit / delta_f_exp_fit if delta_f_exp_fit > 0 else np.nan

        print(f"\n--- 実測データ フィット結果 ---")
        print(f"初期パラメータ (推定): A={initial_A_fit:.4f}, x0={initial_x0_fit:.2f}, gamma={initial_gamma_fit:.2f}, B={initial_B_fit:.4f}")
        print(f"フィットされたパラメータ: A={fit_A:.4f}, x0={fit_x0:.2f}, gamma={fit_gamma:.2f}, B={fit_B:.4f}")
        print(f"フィットされた共振周波数 (f0_exp_fit): {f0_exp_fit:.2f} Hz")
        print(f"フィットされた帯域幅 (Δf_exp_fit): {delta_f_exp_fit:.2f} Hz")
        print(f"フィットされたQ値 (Q_exp_fit): {Q_exp_fit:.2f}")

        fit_success = True
    except (RuntimeError, ValueError) as e:
        print(f"\n--- 実測データ フィット失敗 ---")
        print(f"エラー: {e}")
        print("初期パラメータの調整やフィット範囲の再検討が必要かもしれません。")

# --- グラフの描画 ---
plt.figure(figsize=(12, 7))

# グラフ1: 実測データの散布図
plt.scatter(exp_frequencies, exp_gains, color='blue', marker='o', label='測定値', zorder=5) # zorderで手前に表示

# フィット曲線があればプロット
if fit_success:
    plt.plot(plot_fit_frequencies, fit_gains_for_plot, color='purple', linestyle='--', linewidth=2, label='実測フィット曲線', zorder=4)
    # plt.axvline(x=f0_exp_fit, color='lightblue', linestyle='--', label=f'実測フィット $f_0 = {f0_exp_fit/1000:.2f}$ kHz', zorder=3) # フィットからのf0

# グラフ2: 理論値の線グラフ (スケール調整後)
plt.plot(theory_frequencies, scaled_theory_gains, color='red', linestyle='-', linewidth=2, label='理論値', zorder=2)

# --- テキスト表示位置の調整 ---
x_right_pos = max_freq_val * 0.9
x_left_pos = min_freq_val * 1.5

# 理論共振周波数の値を右に寄せる (削除)
# plt.text(x_right_pos, np.max(scaled_theory_gains) * 0.8, f'理論 $f_0$={f0_theory/1000:.2f}kHz',
#          color='gray', ha='right', va='center', fontproperties=font_prop, fontsize=20)

# 理論Q値のテキストを右に寄せる (削除)
# plt.text(x_right_pos, np.max(scaled_theory_gains) * 0.95, f'理論Q値: {Q_theory:.2f}',
#          color='darkviolet', ha='right', va='top', fontproperties=font_prop, fontsize=20,
#          bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="darkviolet", lw=0.5, alpha=0.8))

# 実測フィットQ値のテキストを左に寄せる (削除)
if fit_success and not np.isnan(Q_exp_fit):
    # plt.text(x_left_pos, np.max(exp_gains) * 0.85, f'実測フィットQ値: {Q_exp_fit:.2f}',
    #          color='darkblue', ha='left', va='top', fontproperties=font_prop, fontsize=20,
    #          bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="darkblue", lw=0.5, alpha=0.8))
    # 実測フィット共振周波数も表示 (削除)
    # plt.text(x_left_pos, np.max(exp_gains) * 0.7, f'実測フィット $f_0$={f0_exp_fit/1000:.2f}kHz',
    #          color='blue', ha='left', va='center', fontproperties=font_prop, fontsize=20)
    pass


# 帯域幅の表示
# 理論値の帯域幅 (削除)
if not np.isnan(delta_f_theory):
    # plt.hlines(y=half_power_gain_theory, xmin=f_low_theory, xmax=f_high_theory, color='green', linestyle=':', label='理論-3dB帯域幅', zorder=1)
    # plt.text((f_low_theory + f_high_theory) / 2, half_power_gain_theory * 1.1, f'理論$\\Delta f$={delta_f_theory/1000:.2f}kHz',
    #          color='green', ha='center', va='bottom', fontproperties=font_prop, fontsize=20)
    pass # 帯域幅のラインとテキストを削除するため、passを記述

# 実測フィットからの帯域幅 (半値点とライン) (削除)
if fit_success and not np.isnan(delta_f_exp_fit):
    # ローレンツ関数のgammaが半値全幅なので、これを使って半値点の高さを計算
    # 半値点は A/2 + B の高さ
    half_power_gain_fit_val = fit_A / 2 + fit_B
    f_low_fit = fit_x0 - fit_gamma / 2
    f_high_fit = fit_x0 + fit_gamma / 2

    # plt.hlines(y=half_power_gain_fit_val, xmin=f_low_fit, xmax=f_high_fit, color='darkorange', linestyle='-.', label='実測フィット-3dB帯域幅', zorder=1)
    # plt.text((f_low_fit + f_high_fit) / 2, half_power_gain_fit_val * 0.9, f'実測フィット$\\Delta f$={delta_f_exp_fit/1000:.2f}kHz',
    #          color='darkorange', ha='center', va='top', fontproperties=font_prop, fontsize=20)
    pass # 帯域幅のラインとテキストを削除するため、passを記述


plt.xscale('log') # 周波数は対数スケールで表示
plt.xlabel("周波数 [Hz]", fontproperties=font_prop, fontsize=20)
plt.ylabel("ゲイン", fontproperties=font_prop, fontsize=20) # ゲインとして統一
plt.title("測定値ゲインとRLC直列共振回路の理論ゲインの比較", fontproperties=font_prop, fontsize=20)
plt.grid(True, which="both", ls="--", c='0.7') # 主グリッドと副グリッドの両方を表示

# y軸の範囲を調整 (両方のデータが収まるように)
max_y_value = max(np.max(exp_gains), np.max(scaled_theory_gains))
plt.ylim(0, max_y_value * 1.1)

plt.legend(prop=font_prop, loc='upper left', fontsize=20) # 凡例のフォントサイズも20に設定
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
from scipy.optimize import curve_fit

# --- 日本語フォント設定 ---
font_path = "C:/Windows/Fonts/meiryo.ttc" # Windowsの場合の例

try:
    font_prop = fm.FontProperties(fname=font_path, size=20) # フォントサイズを20に統一
    plt.rcParams['font.family'] = font_prop.get_name()
    plt.rcParams['axes.unicode_minus'] = False # マイナス記号の表示対応
except FileNotFoundError:
    print(f"警告: 指定されたフォントファイルが見つかりません。{font_path}")
    print("日本語表示が正しく行われない可能性があります。")
    plt.rcParams['font.family'] = 'sans-serif'
    plt.rcParams['font.size'] = 20 # デフォルトフォントサイズも20に設定
    plt.rcParams['axes.unicode_minus'] = False
    pass

# --- グラフ1: 実測データの準備 ---
data = [(10, 0.007895), (100, 0.009375),(1000, 0.012121),(10000, 0.056667),(20000, 0.162931),(21000, 0.258077),(22000, 0.399457),(23000, 0.583333),(24000, 0.608696),(25000, 0.613636),
        (26000, 0.514286),(27000, 0.362745),(28000, 0.253472), (29000, 0.194595), (30000, 0.147826), (31000, 0.133871), (32069, 0.112857), (33000, 0.1), (34063, 0.084375), (35000, 0.078676),
        (36000, 0.08),(37000, 0.077885), (38000, 0.074605), (39000, 0.070707), (40000, 0.065223), (50000, 0.079245), (60000, 0.032258), (70000, 0.024854), (80000, 0.024381), (90000, 0.020841),
        (100000, 0.018519),(500000, 0.00708),(1000000, 0.005941)]

exp_frequencies = np.array([item[0] for item in data]) # numpy配列に変換
exp_gains = np.array([item[1] for item in data])     # numpy配列に変換

# --- グラフ2: 理論値データの準備 (RLC直列共振回路の電流特性) ---
# 定数の定義
R_external = 51.0      # 外部抵抗 [Ω]
R_L_inductor = 29.3    # インダクタンスの抵抗 [Ω]
R_total = R_external + R_L_inductor # 総抵抗 [Ω]

L = 4700e-6 # インダクタンス [H] (4.7 mH)
C = 0.01e-6 # キャパシタンス [F] (0.01 uF)
V = 1.0     # 印加電圧 [V]

# 理論共振周波数f0を計算
f0_theory = 1 / (2 * np.pi * np.sqrt(L * C))

# 理論Q値の計算 (総抵抗を使用)
Q_theory = (2 * np.pi * f0_theory * L) / R_total

print(f"総抵抗 R_total: {R_total:.2f} Ω")
print(f"理論的な共振周波数 f0: {f0_theory:.2f} Hz")
print(f"理論的なQ値: {Q_theory:.2f}")

# 周波数の範囲を生成
min_freq_val = min(np.min(exp_frequencies), f0_theory / 100)
max_freq_val = max(np.max(exp_frequencies), f0_theory * 100)
theory_frequencies = np.logspace(np.log10(min_freq_val), np.log10(max_freq_val), 1000)

# 各周波数でのインピーダンスと電流の計算
theory_currents = []
for f in theory_frequencies:
    omega = 2 * np.pi * f
    XL = omega * L
    XC = 1 / (omega * C)
    
    Z = np.sqrt(R_total**2 + (XL - XC)**2)
    theory_currents.append(V / Z)

max_exp_gain_raw = np.max(exp_gains) # 生の実測ゲインの最大値
max_theory_current = np.max(theory_currents)
scaled_theory_gains = np.array(theory_currents) * (max_exp_gain_raw / max_theory_current)

# --- 理論値の半値全幅 (-3dB点) の計算 (前回と同様) ---
half_power_gain_theory = np.max(scaled_theory_gains) / np.sqrt(2)

peak_idx_theory = np.argmax(scaled_theory_gains)
f_low_theory = np.nan
f_high_theory = np.nan

for i in range(peak_idx_theory -1, -1, -1):
    if scaled_theory_gains[i] <= half_power_gain_theory:
        if i > 0 and scaled_theory_gains[i+1] > half_power_gain_theory:
            f_low_theory = theory_frequencies[i] + (theory_frequencies[i+1] - theory_frequencies[i]) * \
                           (half_power_gain_theory - scaled_theory_gains[i]) / (scaled_theory_gains[i+1] - scaled_theory_gains[i])
            break
        elif i == 0:
            f_low_theory = theory_frequencies[i]
            break

for i in range(peak_idx_theory + 1, len(scaled_theory_gains)):
    if scaled_theory_gains[i] <= half_power_gain_theory:
        if i < len(scaled_theory_gains) -1 and scaled_theory_gains[i-1] > half_power_gain_theory:
            f_high_theory = theory_frequencies[i-1] + (theory_frequencies[i] - theory_frequencies[i-1]) * \
                           (half_power_gain_theory - scaled_theory_gains[i-1]) / (scaled_theory_gains[i] - scaled_theory_gains[i-1])
            break
        elif i == len(scaled_theory_gains) -1:
            f_high_theory = theory_frequencies[i]
            break

delta_f_theory = f_high_theory - f_low_theory if not np.isnan(f_low_theory) and not np.isnan(f_high_theory) else np.nan
print(f"理論的な帯域幅 (Δf_theory): {delta_f_theory:.2f} Hz")


# --- 実測データのフィット (ローレンツ関数) ---

def lorentzian(x, A, x0, gamma, B):
    """
    ローレンツ関数: A * (gamma/2)^2 / ((x - x0)^2 + (gamma/2)^2) + B
    gamma は半値全幅
    """
    return A * ((gamma / 2)**2) / ((x - x0)**2 + (gamma / 2)**2) + B

# 初期パラメータの自動推定の改善
exp_peak_idx_for_fit = np.argmax(exp_gains)
initial_x0_fit = exp_frequencies[exp_peak_idx_for_fit]
initial_max_gain = exp_gains[exp_peak_idx_for_fit]
initial_min_gain = np.min(exp_gains)
initial_A_fit = initial_max_gain - initial_min_gain # ピーク高さの推定
initial_B_fit = initial_min_gain # ベースラインの推定

# 半値全幅の初期推定を、データからもう少し賢く行う
half_gain_for_gamma = initial_max_gain - initial_A_fit / 2 # Aがピーク高さなので、A/2の高さの点を探す
low_freq_gamma_idx = np.where(exp_frequencies < initial_x0_fit)[0]
high_freq_gamma_idx = np.where(exp_frequencies > initial_x0_fit)[0]

f_low_gamma_est = np.nan
if len(low_freq_gamma_idx) > 0:
    for i in low_freq_gamma_idx[::-1]: # ピークから左に
        if exp_gains[i] <= half_gain_for_gamma:
            # 簡易線形補間
            if i + 1 < len(exp_frequencies) and exp_gains[i+1] > half_gain_for_gamma:
                f_low_gamma_est = exp_frequencies[i] + (exp_frequencies[i+1] - exp_frequencies[i]) * \
                                  (half_gain_for_gamma - exp_gains[i]) / (exp_gains[i+1] - exp_gains[i])
            else:
                f_low_gamma_est = exp_frequencies[i]
            break
if np.isnan(f_low_gamma_est):
    f_low_gamma_est = exp_frequencies[0]

f_high_gamma_est = np.nan
if len(high_freq_gamma_idx) > 0:
    for i in high_freq_gamma_idx: # ピークから右に
        if exp_gains[i] <= half_gain_for_gamma:
            # 簡易線形補間
            if i > 0 and exp_gains[i-1] > half_gain_for_gamma:
                f_high_gamma_est = exp_frequencies[i-1] + (exp_frequencies[i] - exp_frequencies[i-1]) * \
                                  (half_gain_for_gamma - exp_gains[i-1]) / (exp_gains[i] - exp_gains[i-1])
            else:
                f_high_gamma_est = exp_frequencies[i]
            break
if np.isnan(f_high_gamma_est):
    f_high_gamma_est = exp_frequencies[-1]

initial_gamma_fit = abs(f_high_gamma_est - f_low_gamma_est) if not np.isnan(f_low_gamma_est) and not np.isnan(f_high_gamma_est) else 5000 # デフォルト値

p0 = [initial_A_fit, initial_x0_fit, initial_gamma_fit, initial_B_fit]

# フィット範囲の限定 (共振周波数の0.5倍から2倍の範囲に限定)
# データ点が多い場合は、ピークの左右の半値点±αの範囲にする
fit_min_freq = initial_x0_fit * 0.5
fit_max_freq = initial_x0_fit * 2.0

fit_indices = np.where((exp_frequencies >= fit_min_freq) & (exp_frequencies <= fit_max_freq))
exp_frequencies_fit_range = exp_frequencies[fit_indices]
exp_gains_fit_range = exp_gains[fit_indices]

# フィットの実行
fit_success = False
if len(exp_frequencies_fit_range) > 3: # 最低4つのパラメータなので4点以上必要
    try:
        # Bの最大値を、観測された最小ゲインの5倍に制限することで、ベースラインのフィットを改善
        bounds = ([0, initial_x0_fit * 0.5, 0, 0], [np.inf, initial_x0_fit * 2, np.inf, initial_min_gain * 5])
        popt, pcov = curve_fit(lorentzian, exp_frequencies_fit_range, exp_gains_fit_range, p0=p0, bounds=bounds, maxfev=10000)
        # 最適化されたパラメータ
        fit_A, fit_x0, fit_gamma, fit_B = popt

        # フィット曲線上のデータを生成
        plot_fit_frequencies = np.logspace(np.log10(min_freq_val), np.log10(max_freq_val), 500)
        fit_gains_for_plot = lorentzian(plot_fit_frequencies, fit_A, fit_x0, fit_gamma, fit_B)

        # フィット結果からの実測Q値と帯域幅の計算
        f0_exp_fit = fit_x0
        delta_f_exp_fit = fit_gamma # ローレンツ関数のgammaが半値全幅
        Q_exp_fit = f0_exp_fit / delta_f_exp_fit if delta_f_exp_fit > 0 else np.nan

        print(f"\n--- 実測データ フィット結果 ---")
        print(f"初期パラメータ (推定): A={initial_A_fit:.4f}, x0={initial_x0_fit:.2f}, gamma={initial_gamma_fit:.2f}, B={initial_B_fit:.4f}")
        print(f"フィットされたパラメータ: A={fit_A:.4f}, x0={fit_x0:.2f}, gamma={fit_gamma:.2f}, B={fit_B:.4f}")
        print(f"フィットされた共振周波数 (f0_exp_fit): {f0_exp_fit:.2f} Hz")
        print(f"フィットされた帯域幅 (Δf_exp_fit): {delta_f_exp_fit:.2f} Hz")
        print(f"フィットされたQ値 (Q_exp_fit): {Q_exp_fit:.2f}")

        fit_success = True
    except (RuntimeError, ValueError) as e:
        print(f"\n--- 実測データ フィット失敗 ---")
        print(f"エラー: {e}")
        print("初期パラメータの調整やフィット範囲の再検討が必要かもしれません。")

# --- グラフの描画 ---
plt.figure(figsize=(10, 6))

# グラフ1: 実測データの散布図
plt.scatter(exp_frequencies, exp_gains, color='blue', marker='o', label='測定値', zorder=5) # zorderで手前に表示

# フィット曲線があればプロット
if fit_success:
    plt.plot(plot_fit_frequencies, fit_gains_for_plot, color='purple', linestyle='--', linewidth=2, label='実測フィット曲線', zorder=4)
    # plt.axvline(x=f0_exp_fit, color='lightblue', linestyle='--', label=f'実測フィット $f_0 = {f0_exp_fit/1000:.2f}$ kHz', zorder=3) # フィットからのf0

# グラフ2: 理論値の線グラフ (スケール調整後)
plt.plot(theory_frequencies, scaled_theory_gains, color='red', linestyle='-', linewidth=2, label='理論値', zorder=2)

# --- テキスト表示位置の調整 ---
x_right_pos = max_freq_val * 0.9
x_left_pos = min_freq_val * 1.5

# 理論共振周波数の値を右に寄せる (削除)
# plt.text(x_right_pos, np.max(scaled_theory_gains) * 0.8, f'理論 $f_0$={f0_theory/1000:.2f}kHz',
#          color='gray', ha='right', va='center', fontproperties=font_prop, fontsize=20)

# 理論Q値のテキストを右に寄せる (削除)
# plt.text(x_right_pos, np.max(scaled_theory_gains) * 0.95, f'理論Q値: {Q_theory:.2f}',
#          color='darkviolet', ha='right', va='top', fontproperties=font_prop, fontsize=20,
#          bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="darkviolet", lw=0.5, alpha=0.8))

# 実測フィットQ値のテキストを左に寄せる (削除)
if fit_success and not np.isnan(Q_exp_fit):
    # plt.text(x_left_pos, np.max(exp_gains) * 0.85, f'実測フィットQ値: {Q_exp_fit:.2f}',
    #          color='darkblue', ha='left', va='top', fontproperties=font_prop, fontsize=20,
    #          bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="darkblue", lw=0.5, alpha=0.8))
    # 実測フィット共振周波数も表示 (削除)
    # plt.text(x_left_pos, np.max(exp_gains) * 0.7, f'実測フィット $f_0$={f0_exp_fit/1000:.2f}kHz',
    #          color='blue', ha='left', va='center', fontproperties=font_prop, fontsize=20)
    pass


# 帯域幅の表示
# 理論値の帯域幅 (削除)
if not np.isnan(delta_f_theory):
    # plt.hlines(y=half_power_gain_theory, xmin=f_low_theory, xmax=f_high_theory, color='green', linestyle=':', label='理論-3dB帯域幅', zorder=1)
    # plt.text((f_low_theory + f_high_theory) / 2, half_power_gain_theory * 1.1, f'理論$\\Delta f$={delta_f_theory/1000:.2f}kHz',
    #          color='green', ha='center', va='bottom', fontproperties=font_prop, fontsize=20)
    pass # 帯域幅のラインとテキストを削除するため、passを記述

# 実測フィットからの帯域幅 (半値点とライン) (削除)
if fit_success and not np.isnan(delta_f_exp_fit):
    # ローレンツ関数のgammaが半値全幅なので、これを使って半値点の高さを計算
    # 半値点は A/2 + B の高さ
    half_power_gain_fit_val = fit_A / 2 + fit_B
    f_low_fit = fit_x0 - fit_gamma / 2
    f_high_fit = fit_x0 + fit_gamma / 2

    # plt.hlines(y=half_power_gain_fit_val, xmin=f_low_fit, xmax=f_high_fit, color='darkorange', linestyle='-.', label='実測フィット-3dB帯 dB幅', zorder=1)
    # plt.text((f_low_fit + f_high_fit) / 2, half_power_gain_fit_val * 0.9, f'実測フィット$\\Delta f$={delta_f_exp_fit/1000:.2f}kHz',
    #          color='darkorange', ha='center', va='top', fontproperties=font_prop, fontsize=20)
    pass # 帯域幅のラインとテキストを削除するため、passを記述


plt.xscale('log') # 周波数は対数スケールで表示
plt.xlabel("周波数 [Hz]", fontproperties=font_prop, fontsize=20)
plt.ylabel("ゲイン", fontproperties=font_prop, fontsize=20) # ゲインとして統一
plt.title("測定値ゲインとRLC直列共振回路の理論ゲインの比較", fontproperties=font_prop, fontsize=20)
plt.grid(True, which="both", ls="--", c='0.7') # 主グリッドと副グリッドの両方を表示

# y軸の範囲を調整 (両方のデータが収まるように)
max_y_value = max(np.max(exp_gains), np.max(scaled_theory_gains))
plt.ylim(0, max_y_value * 1.1)

plt.legend(prop=font_prop, loc='upper left', fontsize=20) # 凡例のフォントサイズも20に設定
plt.tick_params(axis='x', labelsize=20) # x軸の目盛り数字のフォントサイズ
plt.tick_params(axis='y', labelsize=20) # y軸の目盛り数字のフォントサイズ
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

# --- 日本語フォント設定 ---
font_path = "C:/Windows/Fonts/meiryo.ttc" # Windowsの場合の例

try:
    font_prop = fm.FontProperties(fname=font_path, size=12)
    plt.rcParams['font.family'] = font_prop.get_name()
    plt.rcParams['axes.unicode_minus'] = False # マイナス記号の表示対応
except FileNotFoundError:
    print(f"警告: 指定されたフォントファイルが見つかりません。{font_path}")
    print("日本語表示が正しく行われない可能性があります。")
    pass

# --- RLC直列回路の定数 ---
R = 51.0       # 抵抗 [Ω]
L = 0.01e-6    # インダクタンス [H] (0.01 μH)
C = 4700e-6    # キャパシタンス [F] (4700 μF)
V_in = 1.0     # 印加されるステップ電圧の最終値 [V] (例として1V)

# --- 減衰振動のパラメータ計算 ---
# 減衰係数 α (alpha)
alpha = R / (2 * L)

# 固有角周波数 ω0 (omega_0)
omega_0 = 1 / np.sqrt(L * C)

# Q値の計算 (Q = ω0 * L / R)
Q = omega_0 * L / R

print(f"抵抗 R: {R:.1f} Ω")
print(f"インダクタンス L: {L*1e6:.2f} μH")
print(f"キャパシタンス C: {C*1e6:.0f} μF")
print(f"減衰係数 α: {alpha:.2f} s^-1")
print(f"固有角周波数 ω0: {omega_0:.2f} rad/s")
print(f"Q値: {Q:.2f}")

# 減衰振動条件の確認 (アンダーダンプ: α < ω0 または Q > 0.5)
if alpha < omega_0:
    print("回路は減衰振動（アンダーダンプ）モードです。")
    # 減衰角周波数 ωd (omega_d)
    omega_d = np.sqrt(omega_0**2 - alpha**2)
    print(f"減衰角周波数 ωd: {omega_d:.2f} rad/s")
    f_d = omega_d / (2 * np.pi)
    print(f"減衰周波数 fd: {f_d:.2f} Hz ({f_d/1000:.2f} kHz)")
elif alpha == omega_0:
    print("回路は臨界減衰モードです。")
    omega_d = 0
else: # alpha > omega_0
    print("回路は過減衰モードです。")
    omega_d = 0 # 振動しないため

# --- 時間軸の設定 ---
# 減衰時定数 [s]
time_constant = 1 / alpha

# 時間軸の範囲を調整 (非常に大きな時定数になる可能性があるため、動的に調整)
# 過減衰の場合は、振動がないため、時定数を基に収束するまでを見る
if alpha >= omega_0: # 過減衰または臨界減衰
    t_end_calc = time_constant * 5 # 5時定数分
else: # アンダーダンプ
    # 減衰が速いので、数サイクル分 + 減衰が落ち着くまでの時間
    period = (2 * np.pi / omega_d) if omega_d > 0 else (1 / omega_0)
    t_end_calc = max(time_constant * 5, period * 5) # 5時定数と5周期の大きい方

# ただし、今回の定数では時定数が非常に短くなる (alpha が大きいため)
# そのため、時間軸をナノ秒オーダーにする必要があるかもしれません
if t_end_calc < 1e-6: # 1マイクロ秒より短い場合
    t_end_calc = 1e-6 # 少なくとも1マイクロ秒は表示
    time_unit_label = "ナノ秒 [ns]"
    time_scale = 1e9
elif t_end_calc < 1e-3: # 1ミリ秒より短い場合
    time_unit_label = "マイクロ秒 [μs]"
    time_scale = 1e6
else:
    time_unit_label = "ミリ秒 [ms]"
    time_scale = 1e3

t = np.linspace(0, t_end_calc, 1000) # 0からt_end_calcまで1000点

# --- RLC直列回路のステップ応答 (コンデンサ電圧 Vc(t)) ---
# ステップ電圧 V_in が t=0 で印加された場合
if alpha < omega_0: # アンダーダンプの場合
    Vc_t = V_in * (1 - np.exp(-alpha * t) * (np.cos(omega_d * t) + (alpha / omega_d) * np.sin(omega_d * t)))
elif alpha == omega_0: # 臨界減衰
    Vc_t = V_in * (1 - np.exp(-alpha * t) * (1 + alpha * t))
else: # 過減衰
    s1 = -alpha + np.sqrt(alpha**2 - omega_0**2)
    s2 = -alpha - np.sqrt(alpha**2 - omega_0**2)
    # 初期条件 Vc(0)=0, I(0)=0 から係数を導出
    # C1 + C2 = V_in (t=0でVc=0, V_in - (C1+C2) = 0)
    # s1*C1 + s2*C2 = 0 (t=0でI=0, I = C*dVc/dt)
    # これを解くと C1 = V_in * s2 / (s2 - s1), C2 = -V_in * s1 / (s2 - s1)
    # Vc(t) = V_in - C1 * exp(s1*t) - C2 * exp(s2*t)
    
    # 実際は Vc(t) = V_in (1 - A1*exp(s1*t) - A2*exp(s2*t)) の形
    # RLCのステップ応答の一般的な式として
    A1 = (omega_0**2) / (s1 * (s1 - s2))
    A2 = (omega_0**2) / (s2 * (s2 - s1))
    Vc_t = V_in * (1 - A1 * np.exp(s1 * t) - A2 * np.exp(s2 * t))
    print("注意: 過減衰の厳密な波形をプロットしています。")

# --- グラフの描画 ---
plt.figure(figsize=(10, 6))

plt.plot(t * time_scale, Vc_t, label='コンデンサ電圧 $V_C(t)$', color='blue')
plt.axhline(V_in, color='gray', linestyle='--', label=f'最終電圧 $V_{{in}} = {V_in:.1f}$ V')

plt.xlabel(f"時間 [{time_unit_label}]", fontproperties=font_prop)
plt.ylabel("電圧 [V]", fontproperties=font_prop)
plt.title(f"RLC直列回路のステップ応答 (R={R:.0f}Ω, L={L*1e9:.2f}nH, C={C*1e6:.0f}μF)", fontproperties=font_prop)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(prop=font_prop)
plt.xlim(0, t_end_calc * time_scale) # 時間軸を適切に設定
plt.ylim(0, V_in * 1.2) # 少し上まで表示

# --- 減衰時間の表示 ---
print("\n--- 減衰時間に関する情報 ---")
print(f"減衰時定数 (1/α): {time_constant * time_scale:.2f} {time_unit_label.split(' ')[0]}")
print(f"5時定数 (5/α) での減衰時間: {time_constant * 5 * time_scale:.2f} {time_unit_label.split(' ')[0]} (この時間でほぼ最終値に収束します)")

plt.show()

# --- 計算された減衰周波数 ---
if alpha < omega_0:
    print("\n--- 計算された減衰周波数 ---")
    print(f"計算された減衰周波数 fd: {f_d/1000:.2f} kHz")
else:
    print("\n回路は減衰振動しない（過減衰または臨界減衰）ため、減衰周波数は計算されません。")
```

```python
import numpy as np # 数値計算ライブラリ
import matplotlib.pyplot as plt # グラフ描画ライブラリ

# --- 理論曲線と近似曲線のためのパラメータ ---
# これらのパラメータは、画像に示されている共振曲線を視覚的に近似するために選択されています。
# 共振周波数 f0 = 1 / (2 * pi * sqrt(L * C))
# Q値 = (1/R) * sqrt(L/C)

# 目標の共振周波数を約 30 kHz (3e4 Hz) に設定します
f0_target = 3e4  # Hz

# RとQ値を先に決めて、LとCを導出することができます。
R_val = 100  # オーム (任意の値を設定、曲線の形状に合わせて調整)
Q_factor = 50  # Q値 (任意の値を設定、曲線の鋭さに合わせて調整)

# Q = (1/R) * sqrt(L/C) と f0 = 1 / (2 * pi * sqrt(L * C)) から
# L = Q * R / (2 * pi * f0)
# C = 1 / (2 * pi * f0 * Q * R) を導出できます

L_val = Q_factor * R_val / (2 * np.pi * f0_target)
C_val = 1 / (2 * np.pi * f0_target * Q_factor * R_val)

# 近似曲線については、理論曲線と視覚的にわずかに異なるように調整できます。
# （例：わずかに幅を広げたり、ピークをずらしたり）
# 今回のグラフでは理論曲線と近似曲線の違いがごくわずかなので、
# シンプルにするために理論曲線と同じパラメータを使用します。
R_approx = R_val
L_approx = L_val
C_approx = C_val


# --- 曲線をプロットするための周波数範囲 ---
# 10^1 Hz から 10^6 Hz までを500点で対数的に生成します
frequencies = np.logspace(1, 6, 500)
# 角周波数 omega = 2 * pi * f を計算します
omegas = 2 * np.pi * frequencies


# --- 理論曲線と近似曲線のゲインを計算する関数 ---
def calculate_gain(R, L, C, omega):
    # インピーダンスの大きさの計算
    impedance_magnitude = np.sqrt(R**2 + (omega * L - 1 / (omega * C))**2)
    # ゲイン (利得) の計算: R / |Z|
    gain = R / impedance_magnitude
    return gain

# 理論曲線のゲインを計算
theoretical_gain = calculate_gain(R_val, L_val, C_val, omegas)
# 近似曲線のゲインを計算 (今回は理論値と同じパラメータを使用)
approximation_gain = calculate_gain(R_approx, L_approx, C_approx, omegas)


# --- 合成的な実験データを生成 ---
# 画像の実験データの分布とピークに大まかに合わせます。
exp_freq = np.array([
    10, 100, 1000,
    1.5e4, 2e4, 2.5e4, 2.8e4, 3e4, 3.1e4, 3.2e4, 3.3e4, 3.5e4, 3.8e4, 4e4, 4.5e4, 5e4, 6e4,
    7e4, 8e4, 9e4, 1e5, 1.5e5, 2e5, 5e5, 1e6
])

# これらの周波数に対応するゲインの基本値を計算します
exp_omega = 2 * np.pi * exp_freq
exp_gain_base = calculate_gain(R_val, L_val, C_val, exp_omega)

# ランダムなノイズを追加し、適切な範囲に収まるようにします
np.random.seed(42) # 再現性のために乱数シードを設定
noise = np.random.normal(0, 0.02, len(exp_gain_base)) # 平均0、標準偏差0.02の正規分布ノイズ
exp_gain = exp_gain_base + noise
exp_gain[exp_gain < 0] = 0 # ゲインが負にならないように0に設定

# エラーバーをシミュレート (ゲインに比例させるか一定値にするか選択可能)
# ここでは、周波数帯によってエラーバーの大きさを変えています
exp_error = np.where(exp_freq < 1e4, 0.005, np.where(exp_freq < 5e4, 0.02 * exp_gain, 0.005))
exp_error[exp_error > 0.1] = 0.1 # 視認性のため、最大エラーバーを0.1に制限

# 低周波数と高周波数でゲインが低い値になるように調整します
exp_gain[exp_freq < 1000] = 0.005 + np.random.normal(0, 0.002, len(exp_gain[exp_freq < 1000]))
exp_gain[exp_freq > 1e5] = 0.005 + np.random.normal(0, 0.002, len(exp_gain[exp_freq > 1e5]))
exp_gain[exp_gain < 0] = 0


# --- グラフの描画 ---
plt.figure(figsize=(10, 6)) # グラフのサイズを設定

# 理論曲線をプロット
plt.plot(frequencies, theoretical_gain, 'k--', label='理論曲線', linewidth=2)

# 近似曲線をプロット
plt.plot(frequencies, approximation_gain, 'b-', label='近似曲線', linewidth=2, alpha=0.7)

# 実験値をプロット（エラーバー付き）
plt.errorbar(exp_freq, exp_gain, yerr=exp_error, fmt='o', color='red', markersize=5, capsize=3, label='実験値')

# --- グラフのカスタマイズ ---
plt.xscale('log') # X軸を対数スケールに設定
plt.xlabel('周波数 [Hz]', fontsize=12) # X軸ラベル
plt.ylabel('GAIN', fontsize=12) # Y軸ラベル
plt.title('共振回路 (直流抵抗考慮)', fontsize=14) # グラフタイトル
plt.grid(True, which="both", ls="-", alpha=0.7) # 主グリッドと副グリッドの両方を表示
plt.ylim(-0.05, 1.05) # Y軸の表示範囲を設定
plt.xlim(8, 1.2e6) # X軸の表示範囲を画像に合わせて設定

plt.legend(fontsize=10) # 凡例を表示
plt.tight_layout() # レイアウトを自動調整
plt.show() # グラフを表示
```
