```python
import matplotlib.pyplot as plt

#銅の測定値グラフ
x=[1, 2, 3]
y=[0.37429,0.418579611,0.365585480]
y_line = 0.385
plt.axhline(y=y_line, color='green', linestyle='--', label=f'y={y_line}')

yerr = np.array([0.04386293,0.048036242,0.04390]) 
plt.errorbar(x, y, yerr=yerr, fmt='o', ecolor='r', capsize=5)
plt.ylabel('spacing(J/g*K)')
plt.grid(which = "major", axis = "y", alpha = 0.8,linestyle = "--", linewidth = 1)
plt.xticks(range(1, 4)) 
plt.show()
```
