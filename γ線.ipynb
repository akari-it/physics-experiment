```python
import matplotlib.pyplot as plt

# 日本語フォント（Windows）
plt.rcParams["font.family"] = "MS Gothic"

# データ
voltage_cs = [100, 200, 300, 320, 340, 350, 400, 450, 500, 550, 600]
count_cs =   [0,   0,   0,   1593,1667,1677,1779,1670,1752,1684,1765]

voltage_co = [100, 200, 300, 320, 340, 360, 380, 400, 500, 600]
count_co =   [0,   0,   0,   33,  34,  33,  40,  50,  55,  43]

# グラフ作成
plt.figure(figsize=(8,6))

plt.plot(voltage_cs, count_cs, marker='o', linewidth=2, label='Cs-137')
plt.plot(voltage_co, count_co, marker='s', linewidth=2, label='Co-60')

# 軸ラベル
plt.xlabel("Applied Voltage [V]", fontsize=12)
plt.ylabel("Counts / min", fontsize=12)

# タイトル
plt.title("Characteristic Curve of GM Counter", fontsize=14)

# グリッド
plt.grid(True)

# 凡例
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# 日本語フォント（Windows）
plt.rcParams["font.family"] = "MS Gothic"

# --------------------
# セシウム(Cs-137)
# --------------------
voltage_cs = [100, 200, 300, 320, 340, 350, 400, 450, 500, 550, 600]
count_cs   = [0, 0, 0, 1593, 1667, 1677, 1779, 1670, 1752, 1684, 1765]

plt.figure(figsize=(7,5))
plt.plot(voltage_cs, count_cs, marker='o', linewidth=2)
plt.xlabel("Applied Voltage [V]")
plt.ylabel("Counts / min")
plt.title("Characteristic Curve of GM Counter (Cs-137)")
plt.grid(True)
plt.tight_layout()
plt.show()

# --------------------
# コバルト(Co-60)
# --------------------
voltage_co = [100, 200, 300, 320, 340, 360, 380, 400, 500, 600]
count_co   = [0, 0, 0, 33, 34, 33, 40, 50, 55, 43]

plt.figure(figsize=(7,5))
plt.plot(voltage_co, count_co, marker='o', linewidth=2)
plt.xlabel("Applied Voltage [V]")
plt.ylabel("Counts / min")
plt.title("Characteristic Curve of GM Counter (Co-60)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.plot(voltage_cs, count_cs, marker='o', linewidth=2)

# 印加電圧450 Vを示す縦線
plt.axvline(x=450, color='red', linestyle='--', linewidth=2, label='Applied Voltage = 450 V')

plt.xlabel("Applied Voltage [V]")
plt.ylabel("Counts / min")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

plt.figure(figsize=(7,5))
plt.plot(voltage_co, count_co, marker='o', linewidth=2)

# 印加電圧450 Vを示す縦線
plt.axvline(x=450, color='red', linestyle='--', linewidth=2, label='Applied Voltage = 450 V')

plt.xlabel("Applied Voltage [V]")
plt.ylabel("Counts / min")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
