```python
import numpy as np
import matplotlib.pyplot as plt

# パラメータ
A = 1  # 任意の定数。1にすると標準的なスパイラル
theta = np.linspace(0, 4 * np.pi, 1000)  # θの範囲（0〜4π）
r = A * np.exp(theta)

# グラフ描画（極座標）
ax = plt.subplot(111, polar=True)
ax.plot(theta, r)
ax.set_title(r"$r = Ae^\theta$", va='bottom')
plt.show()

```
